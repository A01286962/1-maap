```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

file_path = 'https://raw.githubusercontent.com/tec03/Datasets/0499cdd7fddd3e4c4ca6c50b463c70c5872d62a3/datasets/antropometria-dataset.csv'
df = pd.read_csv(file_path)
df.head()

```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='talla')
plt.title('Box plot de Tallas')
plt.xlabel('Tallas')
plt.ylabel('')

plt.show()
```

```python
# Recargamos el dataset original para restaurar la columna 'talla'
df = pd.read_csv(file_path)

# Creamos una variable nueva para la tabla pivote para no sobrescribir 'df'
df_pivote = pd.pivot_table(df, index='sexo', values='peso', aggfunc='median')
display(df_pivote.head())

# Corregimos el boxplot para graficar 'peso' por 'sexo' de manera correcta
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='sexo', y='peso')
plt.title('Box plot de Peso por Sexo')
plt.xlabel('Sexo 1 , 2 ')
plt.ylabel('Peso (kg)')
plt.show()
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.boxplot(data=df, x='sexo', y='peso')
plt.title('Box plot de Peso por Sexo')
plt.xlabel('Sexo (1, 2')
plt.ylabel('Peso ')

plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='sexo', y='cintura')
plt.title('Box plot de Cintura por Sexo')
plt.xlabel('Sexo (1, 2')
plt.ylabel('Cintura')
plt.tight_layout()
plt.show()
```

```python
cintura_pivot = pd.pivot_table(
    df,
    values='cintura',
    index='desc_ent',
    columns='sexo',
    aggfunc='median'
)

# 3. Graficar la línea correspondiente al Sexo 1
plt.figure(figsize=(8, 5))
sns.set_theme(style='darkgrid')
sns.lineplot(data=cintura_pivot[1].values)

plt.title('Cintura de Sexo 1')
plt.xlabel('Index')
plt.ylabel('Cintura')
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(data=df, x='peso', hue='sexo', multiple='stack')
plt.title('Histograma de Peso por Sexo')
plt.xlabel('Peso (kg)')
plt.ylabel('Frecuencia')

plt.subplot(1, 2, 2)
sns.histplot(data=df, x='cintura', hue='sexo', multiple='stack')
plt.title('Histograma de Cintura por Sexo')
plt.xlabel('Cintura (cm)')
plt.ylabel('Frecuencia')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x='peso', y='cintura', hue='sexo', alpha=0.7)
plt.title('Relación entre Peso y Cintura')
plt.xlabel('Peso (kg)')
plt.ylabel('Cintura (cm)')
plt.show()
```
