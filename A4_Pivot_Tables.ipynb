```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path_file = 'https://raw.githubusercontent.com/tec03/Datasets/0499cdd7fddd3e4c4ca6c50b463c70c5872d62a3/datasets/adult.csv'
df = pd.read_csv(path_file)
df.head()
```

```python
pivot_adult = pd.pivot_table(df, values=['hours-per-week','age'], index=['marital-status', 'education'], aggfunc='mean')

display(pivot_adult.xs(' HS-grad', level='education'))
```
